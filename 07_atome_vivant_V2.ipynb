# 07 — L'atome vivant du dunun (V2)

Le micro-timing rendu **visible et vivant**, en polaire. Chaque rythme devient un « atome » :
- **angle** = position dans le cycle (12 h) ;
- **rayon** = micro-timing : frappe **en avance → vers l'intérieur**, **en retard → vers l'extérieur** ;
- **flou d'une orbitale** = sa précision ;
- **deux couches** : D1 (externe) + D2 (interne).

La tête de lecture tourne au tempo réel, chaque orbitale **pulse** quand elle sonne, on **sélectionne** les dununs à entendre, et **clic = son**. La *forme* de l'atome = la nature propre du rythme.
