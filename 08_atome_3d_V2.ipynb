# 08 — La tour 3D du dunun (V2)

Le micro-timing en **3D dynamique**. Chaque **cycle** est un étage empilé (le temps monte) → une **tour** qu'on fait tourner à la souris.
- **hauteur** = temps (les étages se resserrent quand ça accélère) ;
- **angle** = position dans le cycle ;
- **rayon** = micro-timing (avance → intérieur, retard → extérieur) ;
- **couleur** = D1 (bleu) / D2 (vert), sélectionnables.

Une tête de lecture (anneau) **monte la tour au tempo réel**, les frappes s'allument, **clic = son**. Projection 3D faite main en Canvas (aucune librairie externe).
