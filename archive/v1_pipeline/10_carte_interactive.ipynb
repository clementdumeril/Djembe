# 10 — CARTE INTERACTIVE : la visu du reel, avec le son 🐦→🥁

L'idée d'origine du projet : une carte où chaque point est un son, et où **cliquer = entendre**.

On génère une page **HTML autonome** (aucun serveur, marche en double-clic) :
- chaque point = une vraie frappe de jembé du corpus (450 frappes, clips embarqués en base64),
- position = ressemblance de timbre (UMAP),
- couleur = type de frappe (bass / tone / slap),
- **survol** = info (type, pièce, rôle), **clic** = le son, légende cliquable pour filtrer.

Le code de génération vit dans `scripts_carte.py` (réutilisable).

```python
import warnings; warnings.filterwarnings('ignore')
from scripts_carte import build_carte
path, n = build_carte()          # ~1 min
import os
print(f'Généré : {path} ({n} frappes, {os.path.getsize(path)/1e6:.1f} Mo)')
```

## Ouvrir la carte
Double-clique sur `output/carte_interactive.html` dans l'explorateur — ou exécute la cellule suivante pour l'ouvrir dans ton navigateur.

```python
import webbrowser, os
webbrowser.open('file://' + os.path.abspath('output/carte_interactive.html'))
```
